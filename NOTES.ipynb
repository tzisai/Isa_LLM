```python

# Define an embedding layer:
# 1000 = total vocabulary/items (num_embeddings)
# 64   = vector dimension size (embedding_dim)
embedding = torch.nn.Embedding(num_embeddings=100, embedding_dim=64)

# Pass discrete integer IDs (e.g., token IDs 3, 42, 99)
input_ids = torch.tensor([3, 42, 99])
embedded_output = embedding(input_ids)

print(embedded_output.shape)
# Output: torch.Size([3, 64]) -> 3 items, each represented by a 64-dim vector
```
